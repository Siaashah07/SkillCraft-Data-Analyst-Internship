```python
import pandas as pd
```

```python
df = pd.read_excel("SkillCraft_Task2_Cleaned.xlsx")
df.head()
```

```python
df[['Sales', 'Quantity', 'Discount', 'Profit']].describe()
```

```python
df.groupby('Category')[['Sales', 'Profit']].sum().sort_values('Profit', ascending=False)
```

```python
df.groupby('Region')[['Sales', 'Profit']].sum().sort_values('Profit', ascending=False)
```

```python
df.groupby('Sub-Category')[['Sales', 'Profit']].sum().sort_values('Profit', ascending=False)
```

```python
df[['Discount', 'Profit']].corr()
```

```python
df[['Sales', 'Profit']].sum()
```
